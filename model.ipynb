### 필요한 모듈 가져오기

```python
import pandas as pd
import numpy as np
import json
import random
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, f1_score
from lightgbm import LGBMClassifier
import joblib
```

### seed 값 고정

```python
# 시드값이 고정이 되어 있지 않을 시 시드 값을 고정하여 모델을 학습 후 평가
# torch, tensorflow 등 사용 시에도 동일한 SEED 값 설정
SEED = 42
random.seed(SEED) 
np.random.seed(SEED)
```

### 데이터셋 불러오기

```python
df = pd.read_csv('./train.csv')
```

```python
df.head()
```

### 전처리

```python
# JSON 형식의 문자열을 리스트로 변환

df['length_sequence'] = df['length_sequence'].apply(json.loads)
df['direction_sequence'] = df['direction_sequence'].apply(json.loads)
df['time_sequence'] = df['time_sequence'].apply(json.loads)
```

```python
# 입력 데이터에 5튜플에 대한 정보를 추가X

def preprocess(df, max_len=40):
    # 필요한 전처리 작업 수행
    inputs = []
    
    for _, row in df.iterrows():
        lengths = row['length_sequence']
        dirs = row['direction_sequence']
        times = row['time_sequence']

        num_packets = len(lengths)

        # 1) base length sequence 확장 (40)
        if len(lengths) >= max_len:
            base = lengths[:max_len]
        else:
            base = lengths + [3000] * (max_len - len(lengths))

        # 2) Length sequence 통계 특징
        if num_packets > 0:
            avg_pkt = np.mean(lengths)
            max_pkt = np.max(lengths)
            std_pkt = np.std(lengths)
            median_pkt = np.median(lengths)
            q1 = np.percentile(lengths, 25)
            q3 = np.percentile(lengths, 75)
        else:
            avg_pkt = max_pkt = std_pkt = median_pkt = q1 = q3 = 0.0

        # 3) Direction 특징
        num_forward = dirs.count(1)
        num_backward = dirs.count(-1)
        dir_change = sum(1 for i in range(len(dirs)-1) if dirs[i] != dirs[i+1])
        ratio_fb = num_forward / (num_backward + 1e-6)

        # 4) IAT 특징
        if len(times) > 1:
            iats = [times[i+1] - times[i] for i in range(len(times)-1)]
            avg_iat = np.mean(iats)
            std_iat = np.std(iats)
            min_iat = np.min(iats)
            max_iat = np.max(iats)
        else:
            avg_iat = std_iat = min_iat = max_iat = 0.0

        # 최종 feature vector
        features = base + [
            num_packets, avg_pkt, max_pkt, median_pkt, std_pkt, q1, q3,
            num_forward, num_backward, ratio_fb, dir_change,
            avg_iat, std_iat, min_iat, max_iat
        ]

        inputs.append(features)

    return inputs
```

```python
# 전처리한 데이터를 열에 추가

df['input'] = preprocess(df)
```

### 학습, 검증 데이터 나누기

```python
# 학습과 검증 데이터 분리
X_train, X_valid, y_train, y_valid = train_test_split(
    df['input'],
    df['label'],
    test_size=0.2,
    random_state=SEED,
    stratify=df['label']
)

# 모델 입력을 위해 리스트 형태로 변환
X_train = X_train.tolist()
X_valid = X_valid.tolist()
y_train = y_train.tolist()
y_valid = y_valid.tolist()
```

### 모델 학습

```python
model = LGBMClassifier(
    n_estimators=500,
    num_leaves=16,
    min_child_samples=50, 
    learning_rate=0.05,
    subsample=0.9,
    colsample_bytree=0.8,
    random_state=SEED,
    n_jobs=-1,
    class_weight='balanced',
    verbosity=-1
)

model.fit(X_train, y_train)
```

### 모델 검증

```python
pred = model.predict(X_valid)
print(f"Accuracy: {accuracy_score(y_valid, pred):.4f}")
print(f"F1 Score: {f1_score(y_valid, pred, average='macro'):.4f}")
```
